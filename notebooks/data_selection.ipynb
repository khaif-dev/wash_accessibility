### Data Filtering
What this notebook does
- import country wide admin (county, constituencies and wards) and osm road data
- filter data to migori county
- export the data as gpkg for pushing to remote repo

```python
# add libraries
from pathlib import Path
import geopandas as gpd
import pandas as pd
```

```python
# config working directory
data_dir = Path.cwd().parent/'raw_datasets'
data_dir
output_dir = Path.cwd().parent/'processed_datasets'
output_dir.mkdir(exist_ok=True, parents=True)
```

```python
# import admin data
adm1 = gpd.read_file(data_dir/'ken_admin_boundaries/ken_admin1.shp')
adm2 = gpd.read_file(data_dir/'ken_admin_boundaries/ken_admin2.shp')
adm4 = gpd.read_file(data_dir/'Kenya_Wards/kenya_wards.shp')
adm4.head(2)
```

```python
# select migori dataset
Migori = adm1[adm1['adm1_name'] == 'Migori']
constituencies = adm2[adm2['adm1_name'] == 'Migori'] 
wards = adm4[adm4['county']== 'Migori']
wards
```

```python
# export to gpkg
Migori.to_file(output_dir/'Migori.gpkg', driver='GPKG')
constituencies.to_file(output_dir/'Migori_const.gpkg', driver='GPKG')
wards.to_file(output_dir/'Migori_wards.gpkg', driver='GPKG')
```

```python
# import road data
roads = gpd.read_file(data_dir/'hotosm_ken_roads_osm/roads_lines.shp')
roads.head(2)
```

```python
# filter to migori and export to gpkg
migori_roads = roads[roads['adm1_name'] == 'Migori']
migori_roads.to_file(output_dir/'roads.gpkg', driver='GPKG')
```

```python
# wash data
wash = pd.read_csv(data_dir/'kenyaWASH.csv')
wash.head(2)
```

```python
wash.columns
```

```python
# filter to migori and export
wash_migori = wash[wash['clean_adm1'] == 'Migori']
wash_migori.to_csv(output_dir/'wash.csv', index=False)
```
